```python
name=input("Enter the Name of the Person : " )
x=input("Enter the Age of the person : " )
print("My Name is {} and my Age is {} ".format(name,x))
#old school method
```

```python
#modern method
#f-strings
print(f"My Name is {name} and my Age is {x} ")
```

```python
#we can do multiple operation using f stings 
a=int(input("Enter the first value of the Person : " ))
b=int(input("Enter the second value of the Person : " ))
sum=print(f"The Sum of {a} and {b} is {a**b} ")
```

```python
#we can do formatting also with help of f string that is rounding up
sum=30.404054332
print(f"The value of sum is upto 2 decimal is :{sum:.2f}")
```

```python
#we can use to align also
print(f"{'Krishna':<11}")#left align
print(f"{'Krishna':>13}")#Right align
print(f"{'Krishna':^32}")#center align
```
