```python
#break statemnts(stop the iteration there if imposter found)
l=[1,3,5,9,8,5,3,1]
for i in l:
    if (i%2==0):
        break
    print(i)
        
```

```python
#continue statements(jump of that imposter and continue the loop)
#break statemnts
l=[1,3,5,9,8,5,3,1]
for i in l:
    if (i%2==0):
        continue
    print(i)
        
```
