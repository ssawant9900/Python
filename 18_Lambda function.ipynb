```python
str=(input("Enter the number of integers : " ))
numbers=list(map(int,str.split()))
evens = list(filter(lambda x: x % 2 == 0, numbers))
div_5 = list(filter(lambda x: x % 5 == 0, numbers))
odd=list(filter(lambda x:x%2==1,numbers))
print(f" The Even integers are {evens}") 
print(f" The Odd integers are {odd}")
print(f" The number which are divisible by 5 are {div_5}")
```

```python
str=input("Enter the Data to segregate greater or lesser number than 50 : " )
num=list(map(int, str.split()))
greater_50=list(filter(lambda x:x>50,num))
lesser_50=list(filter(lambda x:x<50,num))
choice=input("Please Enter Greater or Lesser : ")

if choice=="Greater" :
    print(f"The Numbers Greater than 50 are : {greater_50}")
else :
    print(f"The Numbers lesser than 50 are : {lesser_50}")
```

```python
str=input("Enter the Data to segregate number that are divisble by 3 and 5  : " )
num=list(map(int,str.split()))
div_3=list(filter(lambda x:x%3==0,num))
div_5=list(filter(lambda x:x%5==0,num))
residue = list(filter(lambda x: x % 3 != 0 and x % 5 != 0, num))
print(f"The Number which are divisible by 3 are : {div_3} ")
print(f"The Number which are divisible by 5 are : {div_5} ")

print(f"The Number which are not either divisible by 5 nor by 3 are : {residue} ")
```
