# CLASS Method()

```python
Class method reqiure cls parameter and can acess class Attributes only , it cannot acess instance attribute 
but with object we can acess class attributes conveniently .
we use Decorator function as @classmethod which denotes it is class method and
changes the function behviour it makes that function a class method .
```

```python
class laptop : 
    storage_type="SSD"
    
    def __init__(self,ram,storage):
        self.ram=ram
        self.storage=storage
   
    @classmethod
    def get_type_storage(cls):
        print(f"The Storage type is {cls.storage_type}")
        
    def get_info(self):#Instance method
        print(f'The laptop has {self.ram}RAM & {self.storage} Storage having storage type {self.storage_type}')
```

```python
lap1.get_type_storage()
```

# STATIC Method()

```python
class laptop : 
    storage_type="SSD"
    
    def __init__(self,ram,storage):
        self.ram=ram
        self.storage=storage
   
    @classmethod
    def get_type_storage(cls):
        print(f"The Storage type is {cls.storage_type}")

    @staticmethod
    def get_discount(price,discount):
        final_price=price-(discount*price/100)
        print(f"The discounted price is {final_price}")
        
    @staticmethod # this function doesn acces class attributes and instances attribute but present in the class .
    def get_gst_discounted(price,discount,gst):
        discount_price=price-(discount*price/100)
        final_price=discount_price+(discount_price*(gst/100))
        print(f"The final price after discount and GST is {final_price}")
        
    def get_info(self):#Instance method
        print(f'The laptop has {self.ram}RAM & {self.storage} Storage having storage type {self.storage_type}')
```

```python
lap1.get_gst_discounted(100_000,30,18)
```

Creating a small store were different types of E_products are present .

```python
class store :
    Store_name="Reliance_Digital"
    total_inventory=0
    lap_inventory=0
    phone_inventory=0

    def __init__(self,name,types,price):
        self.name=name
        self.type=types
        self.price=price
        store.total_inventory+=1
        if self.type=="phone":
            store.phone_inventory+=1
        else:
            store.lap_inventory+=1
        
    @staticmethod # this function doesn acces class attributes and instances attribute but present in the class .
    def get_gst_discounted(price,discount,gst):
        discount_price=price-(price*discount/100)
        final_price=discount_price+(discount_price*(gst/100))
        print(f"The final price after discount and GST is {final_price}")

    @classmethod 
    def get_total_inventory(cls):
        print(f"The Total Inventory is {cls.total_inventory}")
    @classmethod 
    def get_total_inventory_lap(cls):
        print(f"The Total Inventory is {cls.lap_inventory}")
    @classmethod 
    def get_total_inventory_phone(cls):
        print(f"The Total Inventory is {cls.phone_inventory}")

    def get_info(self):
        print(f"Name of product  :{self.name}")
        print(f"type of electronic :{self.type}")
        print(f"Price  of product :{self.price}")
```

# Encapsulation

```python
class BankAccount :
    def __init__(self,name,balance):
        self.name=name
        self.__balance=balance # protected as single underscore and private has double __score 

    def get_balance(self):#getter
        print(f'The balance is {self.__balance}')

    def set_balance(self,new): #setter 
        self.__balance+=new
        print(f"The New Balance is {set_balance}")
```

# Inheritance

In Inheritance we get properties and attribute from the parents ,
We Reuse Attribute and Methods from the parent Class.


```python
class Employee :
    start_time="9am"
    End_time="5pm"

class Teacher(Employee): # here by giving Parent Class name in the Argument 
       def __init__(self,subject):#we have inherit all Attribue and methods into Teacher class from Employee
           self.subject=subject
```

```python
t1=Teacher("chem")
```

```python
print(f'The t1 teaches the subject {t1.subject} & start_time is {t1.start_time}')
```

# Type of Inheritance

1.Single level Inheritance 


```python
# Single level Inheritance 
# were there is Single Parent and Single Child
class Employee :
    start_time="9am"
    End_time="5pm"

class Teacher(Employee): # here by giving Parent Class name in the Argument 
       def __init__(self,subject):#we have inherit all Attribue and methods into Teacher class from Employee
           self.subject=subject
```

2.Mutli-Level Inheritance

In multilevel inheritance child inherites attributes and methods from hierarchy of parents , Grandparent --> Parent --> child

```python
class Employee :
    Start_time="9am"
    End_time="5pm"
class Adminstaff(Employee):
    def __init__(self,role):
        self.role=role
class Accountant(Adminstaff):
    def __init__(self,salary,role):
        super().__init__(role) #here super function invckes the role attribute of the Adminstaff becaues it had to acess first role 
                                # which is inheriting from the parent class
        self.salary=salary
```

```python
t1=Accountant(250_000,"CA")
```

```python
t1.role
```

3.Muliple Inheritance 

Here child Inherites the Attributes and Method from Different Parent

```python
class Teacher:
    def __init__(self, salary):
        self.salary = salary
class Student:
    def __init__(self, gpa):
        self.gpa=gpa
class TA(Teacher, Student):
    def __init__(self, salary, gpa, name):
        super().__init__(salary)
        Student.__init__(self, gpa) # while using the class name itself we have to provide self parameter for invoaction of student attribute 
        self.name = name
```

```python
Ta1=TA(50_000,9.4,"Shubham")
Ta1.gpa
```

# Abstraction


Hiding internal details & showing only essential features

we use Abstarct method and decorator and it is blueprint for the other class 

```python
from abc import ABC ,abstractmethod
```

```python
class Animal(ABC):
    @abstractmethod 
    def make_roar():
        pass
        # Here in abstract method we do not use this method in the Abstract class but we use in other class the same function
class Lion(Animal) :
       def make_roar(self):
           print("Roar!!!!!")
class cat(Animal) :
       def make_roar(self):
           print("Meow.....")
```
